```python
import pandas as pd
import random
from datetime import date, timedelta

random.seed(42)

# 1. Generate patient data
patients = []

for i in range(1, 101):
    patients.append({
        "patient_id": f"P{i:04d}",
        "age_band": random.choice(["18-30", "31-45", "46-60", "61-75"]),
        "area_code": random.choice(["NG1", "NG2", "NG5", "NG7"])
    })

patients_df = pd.DataFrame(patients)

# 2. Generate referral data
referrals = []

for i in range(1, 101):
    referrals.append({
        "referral_id": f"R{i:04d}",
        "patient_id": f"P{i:04d}",
        "referral_date": (
            date.today() - timedelta(days=random.randint(5, 180))
        ).isoformat(),
        "specialty": random.choice(
            ["Cardiology", "Orthopaedics", "Dermatology"]
        ),
        "priority": random.choice(["Routine", "Urgent"]),
        "referral_status": "Open"
    })

referrals_df = pd.DataFrame(referrals)

# 3. Generate appointment data
appointments = []

for i in range(1, 101):
    appointments.append({
        "appointment_id": f"A{i:04d}",
        "referral_id": f"R{i:04d}",
        "appointment_date": (
            date.today() + timedelta(days=random.randint(1, 60))
        ).isoformat(),
        "location_id": random.choice(["L001", "L002", "L003"]),
        "appointment_status": random.choice(
            ["Booked", "Attended", "Cancelled"]
        )
    })

appointments_df = pd.DataFrame(appointments)

# 4. Generate clinic capacity data
capacity_df = pd.DataFrame([
    {"location_id": "L001", "location_name": "Central Clinic",
     "specialty": "Cardiology", "available_slots": 5},
    {"location_id": "L002", "location_name": "North Clinic",
     "specialty": "Orthopaedics", "available_slots": 8},
    {"location_id": "L003", "location_name": "South Clinic",
     "specialty": "Dermatology", "available_slots": 6}
])

print("Synthetic data generated successfully")

print("Patients:", len(patients_df))
print("Referrals:", len(referrals_df))
print("Appointments:", len(appointments_df))
print("Capacity:", len(capacity_df))

display(patients_df.head())
display(referrals_df.head())
display(appointments_df.head())
display(capacity_df)



# 5. Generate synthetic unstructured referral documents

referral_notes = [
    "Patient reports recurring chest discomfort and breathlessness.",
    "Patient experiences persistent knee pain and difficulty walking.",
    "Patient reports a recurring skin rash and irritation.",
    "Patient has experienced occasional heart palpitations.",
    "Patient reports joint stiffness and reduced mobility.",
    "Patient has persistent skin inflammation requiring specialist review.",
    "Patient reports shortness of breath during physical activity.",
    "Patient experiences swelling and discomfort in the knee.",
    "Patient reports an ongoing skin condition that has not improved."
]

referral_documents = []

for i in range(1, 101):
    referral_documents.append({
        "document_id": f"D{i:04d}",
        "referral_id": f"R{i:04d}",
        "patient_id": f"P{i:04d}",
        "document_type": "Referral Notes",
        "document_text": random.choice(referral_notes),
        "document_date": (
            date.today() - timedelta(days=random.randint(5, 180))
        ).isoformat()
    })

referral_documents_df = pd.DataFrame(referral_documents)

print("Referral documents:", len(referral_documents_df))
display(referral_documents_df.head())
```

```python
output_path = "/Volumes/healthcare_demo/bronze/raw"

# Convert Pandas DataFrames to Spark DataFrames and write CSV files
spark.createDataFrame(patients_df).coalesce(1).write.mode("overwrite").option("header", True).csv(output_path + "/patients")

spark.createDataFrame(referrals_df).coalesce(1).write.mode("overwrite").option("header", True).csv(output_path + "/referrals")

spark.createDataFrame(appointments_df).coalesce(1).write.mode("overwrite").option("header", True).csv(output_path + "/appointments")

spark.createDataFrame(capacity_df).coalesce(1).write.mode("overwrite").option("header", True).csv(output_path + "/capacity")

# Save unstructured referral documents
spark.createDataFrame(referral_documents_df).coalesce(1).write.mode("overwrite").option("header", True).csv(output_path + "/referral_documents")

print("All synthetic data saved successfully")


```
